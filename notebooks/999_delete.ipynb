# Remove directories in parallel

```python
import os,sys

sys.path.append(os.path.abspath('../'))
import subprocess
from pathlib import Path
from src.config import BASE_PATH, SEED
```

```python
def create_cmd(perm_dir):
    cmd_str = f"export PYTHONPATH={BASE_PATH}; \
			uv run python -m src.remove_dir \
			--dir_path {perm_dir}"
    return " ".join(cmd_str.split())
```

```python
loop_dir = BASE_PATH / "perm_res" / "preds"
swarm_path = BASE_PATH / "swarm_dir" / "delete" / "command.swarm"
if swarm_path.exists():
    swarm_path.unlink()
swarm_path.parent.mkdir(exist_ok=True, parents=True)
cmd_list = []
for perm_dir in loop_dir.iterdir():
    cmd = create_cmd(perm_dir)
    cmd_list.append(cmd)
len(cmd_list)
```

```python
cmd_list_brief = cmd_list
swarm_path.write_text("\n".join(cmd_list_brief))
len(cmd_list_brief)
```

```python
swarm_time = "0:20:00"
num_cpu=4
log_dir =  BASE_PATH / "swarm_dir" / "delete"
n_batch = 1
gb= 2
swarm_cmd = f"swarm --time={swarm_time} -g {gb} -t {num_cpu} --logdir={log_dir}  --job-name=remove --partition=quick -b {n_batch} {swarm_path}"
swarm_cmd
```

```python
subprocess.run(swarm_cmd, shell=True)
```
